```python
import matplotlib.pyplot as plt
import cv2
# reading the image using open cv or using pillow
image = cv2.imread("D:\Data Science by SRK\Omkar Sir 10.30AM Batch\OpenCV\IMG_5749.JPG")

# opening the image using matplotlib
image[1100:1750,1300:1800] = (0,255,0)  # It will make the selected specific part as green color, because image is in BGR format.
plt.imshow(image)

```

```python
#cv2.rectangle(start_point,end_point,color,thickness)

pt1 = (1100,1750)  # (y1,y2)   height = y2-y1
pt2 = (1300,1800)  # (x1,x2)   width = x2-x1
rgb = (255,0,0) # It will show red color on selected part
thickness = 3

plt.imshow(cv2.rectangle(image,pt1,pt2,rgb,thickness))
```
